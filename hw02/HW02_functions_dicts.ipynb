````python
def to_num(s):
    return float(s.replace(" ", ""))

def clean_city(s):
    s = s.strip().lower()

    if s == "almaty":
        return "Алматы"
    elif s == "shymkent":
        return "Шымкент"
    elif s == "astana":
        return "Астана"
    elif s == "karaganda":
        return "Караганда"
    elif s == "aktobe":
        return "Актобе"
    elif s == "atyrau":
        return "Атырау"
    else:
        return "Неизвестный город"


print(to_num("15990"))
print(clean_city("  ALMATY  "))
print(clean_city("SHYMKENT"))
print(clean_city("London"))
````
