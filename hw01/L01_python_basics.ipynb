## 5. Списки: первая «таблица»

Наш сквозной кейс на модуль: **заказы интернет-магазина**. Пока храним колонки в списках, через 3 недели это будет DataFrame в Pandas.

```python
cities  = ["Almaty", "Astana", "Shymkent", "Almaty", "Karaganda", "Astana", "Almaty", "Aktobe"]
amounts = [15990, 4500, 32000, 870, 12500, 56000, 2300, 9900]
paid    = [True, True, False, True, True, False, True, True]

print(len(amounts))
print(amounts[0], amounts[-1])
print(amounts[2:5])
print(sum(amounts), min(amounts), max(amounts))
print(sum(amounts) / len(amounts))   # средний чек
```

## 6. Условия

## 7. Циклы

```python
for a in amounts:
    print(a, end=" ")
print()

for i in range(3):
    print("итерация", i)

# индекс + значение
for i, c in enumerate(cities):
    print(i, c)
```

```python
# zip: идём по нескольким «колонкам» одновременно, как по строкам таблицы
for c, a, p in zip(cities, amounts, paid):
    status = "оплачен" if p else "НЕ оплачен"
    print(f"{c:<10} {a:>7} ₸  {status}")
```

### Паттерн «накопитель»: так считаются почти все метрики

```python
total_paid = 0
n_paid = 0
for a, p in zip(amounts, paid):
    if p:
        total_paid += a
        n_paid += 1
print("Выручка по оплаченным:", total_paid)
print("Средний чек оплаченных:", round(total_paid / n_paid))
```

```python
# while: пока условие истинно
balance = 100000
months = 0
while balance > 0:
    balance -= 23000
    months += 1
print(f"Денег хватит на {months} мес.")
```
