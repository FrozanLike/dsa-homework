````python
#8. Доля заказов выходного дня (`сб`, `вс`) в процентах.
weekend = 0
for i in range(len(days)):
    if days[i] == "сб" or days[i] == "вс":
        weekend = weekend+1
        
print("Общая доля выходных дней в процентах составляет: ", (weekend/len(days)*100), "%")
````

````python
#9. Задача «копилка»: студент откладывает 15 000 ₸ в месяц, курс стоит 400 000 ₸. С помощью `while` посчитайте, 
#через сколько месяцев он накопит. 
#А если каждый месяц откладывать на 1 000 ₸ больше, чем в прошлый?

month = 0
bill = 0
course = 400000
while bill < course:
    month = month + 1
    bill = bill + 15000
print("Студент накопит на курса через ", month, "месяцев")
````

13. *Подумать:* какие проблемы данных вы встретили (пробелы, регистр, числа-строки)? Опишите 2–3 предложениями в Markdown-ячейке. Именно это мы будем делать в Pandas на занятии 6.

````python
unique_cities = []

for i in cities_clean:
    if i not in unique_cities:
        unique_cities.append(i)

print(unique_cities)
````

````python
revenue_by_city = []

for city in unique_cities:
    total = 0

    for i in range(len(cities_clean)):
        if cities_clean[i] == city and paid[i] == True:
            total = total + amounts_int[i]

    revenue_by_city.append(total)

check = True

while check:
    check = False

    for i in range(len(revenue_by_city) - 1):
        if revenue_by_city[i] < revenue_by_city[i + 1]:

            revenue_by_city[i], revenue_by_city[i + 1] = revenue_by_city[i + 1], revenue_by_city[i]

            unique_cities[i], unique_cities[i + 1] = unique_cities[i + 1], unique_cities[i]

            check = True

for city, revenue in zip(unique_cities, revenue_by_city):
    print(f"{city} | {revenue} ₸")
````
